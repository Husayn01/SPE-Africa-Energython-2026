# 00 — Project Framing — Team Artemis

What this section covers: We define the project at a glance — what we are building, where, for how long, and how we will judge success. This page functions as the cover page and executive framing for the whole submission.

**Acronyms first used on this page (expand on first mention for readability):**
- **IT** = Information Technology — the servers inside the data center
- **MW** = Megawatt = 1,000,000 watts of electrical power
- **CCHP** = Combined Cooling, Heat and Power — gas engines make electricity *plus* their waste heat drives free cooling for the servers
- **N+2** = Redundancy rule: install 2 extra engines beyond IT load so 1 engine can be offline without service loss
- **IRR** = Internal Rate of Return — annualized profit rate, analogous to an interest rate on the investment
- **MOIC** = Multiple on Invested Capital — dollars returned per dollar invested by the owners
- **DSCR** = Debt Service Coverage Ratio — quarterly bank covenant: does operating cash cover debt amortization plus interest? Target band: 1.30x to 1.50x.
- **LLCR** = Loan Life Coverage Ratio — total cash available over the entire loan life divided by total outstanding debt; target at least 1.40x at close.
- **FZE** = Free Zone Enterprise — a company licensed inside Lekki Free Zone; qualifies for 0% tax holiday for 5 years (renewable for a further 3 years) plus duty-free capital imports and unrestricted profit repatriation.
- **NPV** = Net Present Value — value in today dollars of all future profits minus all future costs, discounted at the project hurdle rate.
- **LiBr** = Lithium Bromide absorption chiller — uses heat, not electricity, to produce chilled water; central to our CCHP cooling loop.
- **ELPS** = Escravos-Lagos Pipeline System — 439 km, 36-inch gas trunk line running Escravos to Warri to Egbin to Lagos; 800 MMscf/d aggregate capacity.
