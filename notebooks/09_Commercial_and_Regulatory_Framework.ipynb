# Notebook 9: Commercial and Regulatory Framework

## 9.0 Framing: From Selection to Execution
Following the technical design (Notebook 4) and financial validation (Notebook 7), **Location 3** was selected as the optimal site for Project Artemis (Notebook 8). However, a bankable project requires more than strong unit economics; it requires a bulletproof commercial and regulatory framework.

This notebook outlines the legal agreements, licenses, and community frameworks necessary to transition Project Artemis from a "selected site" to a "shovel-ready" investment. It addresses critical dependencies across gas supply, power generation, and community relations.


## 9.1 Gas Supply Framework (NGFCP)

Project Artemis secures its primary fuel through the **Nigerian Gas Flare Commercialisation Programme (NGFCP)**. Administered by the Nigerian Upstream Petroleum Regulatory Commission (NUPRC), this framework requires a specific suite of condition-precedent agreements.

### 9.1.1 Required Agreements
*   **Gas Sales Agreement (GSA):** Executed with the Federal Government of Nigeria (via NUPRC) to transfer legal title of the flare gas to the Artemis project vehicle.
*   **Milestone Development Agreement (MDA):** A commitment to the NUPRC detailing the project execution schedule, ensuring the flare site is commercialized within the agreed timeframe.
*   **Connection Agreement:** Negotiated with the Upstream Producer (the operator of the oil field) to establish the physical tie-in points for the gas gathering pipeline at the flare stack.
*   **Deliver-or-Pay Agreement:** A supplementary agreement with the Upstream Producer to guarantee the minimum gas volumes required by the data center, hedging against production fluctuations.

### 9.1.2 Licensing and Data Access
*   **Permit to Access Flare Gas:** The ultimate statutory license granted upon execution of the GSA and MDA, granting Artemis personnel legal right of entry to the flare site.
*   **Data Room & Permitting Fees:** Provisioned in the financial model's OpEx (as `NERC_NGFCP_REG_FEES_USD`) to maintain statutory compliance and access to NGFCP flare gas forecast data.


## 9.2 Power Generation and Regulatory Licensing

As an "Edge Compute" data center operating entirely behind-the-meter (island mode), Project Artemis avoids national grid wheeling charges but falls under specific off-grid IPP regulations.

### 9.2.1 NERC Off-Grid Generation License
Under the *Electric Power Sector Reform Act* and the *NERC Regulations for Independent Power Producers*, any off-grid facility generating more than 1 MW for exclusive use must obtain a formal permit.
*   **Requirement:** Artemis (at 26.4 MW installed capacity) requires a Off-Grid Generation License from the Nigerian Electricity Regulatory Commission (NERC).
*   **Submission Criteria:** This requires the submission of corporate documentation, Environmental and Social Impact Assessment (ESIA) approvals, and single-line technical diagrams detailing the Jenbacher engines and the Battery Energy Storage System (BESS).

### 9.2.2 Environmental & Social Impact Assessment (ESIA)
Given the 21.6 km pipeline routing (Location 3) and the installation of heavy reciprocating engines, full ESIA approval from the Federal Ministry of Environment is a mandatory precursor to site mobilization.


## 9.3 Offtake & Revenue Contracts (PPA)

The financial bankability of Project Artemis rests entirely on the Power Purchase Agreement (PPA) with the anchor data center tenant.

### 9.3.1 PPA Structuring
*   **Tariff Base:** The base tariff of **$0.14/kWh** offers the offtaker a significant discount compared to self-generated diesel (which currently exceeds $0.40/kWh given local pump prices) while securing a high-yield return for Artemis.
*   **Take-or-Pay Capacity Charge:** To protect the debt service (DSCR), the PPA will include a fixed capacity charge that covers the debt obligations and fixed OpEx, regardless of the data center's utilization rate.
*   **Currency Indexation:** To mitigate the extreme foreign exchange (FX) volatility modeled in the Monte Carlo simulations (Notebook 7), the PPA tariff will be USD-indexed, aligning revenues with the USD-denominated CapEx debt.


## 9.4 Community Relations and Security

Operating a 73.4 km pipeline and a high-value data center in the Niger Delta requires robust community integration to mitigate vandalism and security risks.

### 9.4.1 Host Communities Development Trust (HCDT)
*   **Structure:** Artemis will execute a GMoU with the host communities along the pipeline Right-of-Way (ROW) and the final facility site.
*   **Funding:** Modeled within the `GNA_COMMUNITY_USD` OpEx line item, a percentage of project revenues will be committed to a community trust fund, governed jointly by project representatives and local leaders, to fund local infrastructure and secure the project's social license to operate.


## 9.5 Land Acquisition and Right-of-Way (ROW) Permits

### 9.5.1 Site Land Acquisition
Location 3 occupies a 1.50 sqkm footprint within an area classified as sparse shrubland by the ESA WorldCover dataset. Land acquisition will be executed through a combination of direct purchase from the Ologbo community land trustees and a long-term lease agreement (minimum 25 years with renewal options) registered with the Delta State Ministry of Lands, Survey, and Urban Planning. All land transactions will comply with the Land Use Act (1978), requiring Governor's Consent for any assignment or sublease.

### 9.5.2 Pipeline Right-of-Way (ROW)
The 73.4 km gas gathering pipeline traverses multiple communities and local government areas. ROW acquisition will be executed on a strip-by-strip basis, negotiating corridor easements (typically 15-25 meters wide) with each affected community. Compensation will follow the Federal Ministry of Works standard valuation methodology for crops and economic trees. All ROW agreements will be registered as encumbrances on the relevant Certificates of Occupancy to prevent future disputes.


## 9.6 Engineering, Procurement, and Construction (EPC) Contract

The construction of the gas conditioning facility, power island (eight Jenbacher J620 engines), BESS, and data center halls will be executed under a single **Lump Sum Turnkey (LSTK) EPC contract**. This structure transfers schedule and budget risk to the EPC contractor, insulating the project vehicle from cost overruns during the 18-month construction phase.

**Key EPC Provisions:**
*   **Performance Guarantees:** The EPC contractor will guarantee minimum electrical output (26.4 MW gross installed capacity), heat rate, and emissions compliance. Liquidated damages (LDs) will apply for shortfalls against guaranteed performance levels.
*   **Completion Milestones:** Interim milestone payments will be tied to verifiable construction progress (foundation completion, mechanical completion, first gas, commercial operation date) to manage cash flow during the construction period.
*   **Defects Liability Period:** A minimum 24-month defects liability period will be enforced post-COD, during which the EPC contractor remains responsible for rectifying any design or construction defects at no additional cost.


## 9.7 Operations and Maintenance (O&M) Agreement

Post-commissioning, Artemis will engage a specialist O&M contractor under a **Long-Term Service Agreement (LTSA)** with the engine OEM (INNIO Jenbacher) or an accredited third-party operator. The LTSA will cover:

*   **Scheduled Maintenance:** Major overhauls at 60,000 and 80,000 running-hour intervals, including top-end rebuilds, turbocharger replacements, and generator rewinding.
*   **Spare Parts Inventory:** A minimum critical spares package (one complete cylinder head assembly, one turbocharger cartridge, and one starter motor per engine) will be warehoused on-site to minimize unplanned downtime.
*   **Performance Monitoring:** Real-time engine telemetry (exhaust gas temperature, vibration, lube oil quality) will be transmitted to the OEM's remote monitoring center for predictive maintenance analytics.
*   **Availability Guarantee:** The LTSA will include a contractual availability guarantee of 92% or higher, with bonus/penalty mechanisms tied to actual uptime performance.


## 9.8 Insurance and Force Majeure

### 9.8.1 Insurance Program
A comprehensive insurance program, structured in consultation with a specialist energy insurance broker, will be placed prior to Financial Close. The minimum required policies include:

*   **Construction All Risks (CAR) / Erection All Risks (EAR):** Covers physical damage to works, materials, and equipment during the construction phase, including transit insurance for engine and BESS shipments from European and Asian ports of origin.
*   **Delay in Start-Up (DSU):** Covers lost revenue and debt service obligations in the event that the Commercial Operation Date (COD) is delayed beyond the scheduled 18-month construction window.
*   **Property All Risks / Business Interruption:** Covers physical damage to the operating facility and the consequential loss of revenue during the repair or replacement period.
*   **Third-Party Liability:** Covers claims arising from pipeline leaks, environmental contamination, or bodily injury to community members or contractor personnel.

### 9.8.2 Force Majeure Provisions
All project agreements (GSA, PPA, EPC, LTSA) will contain standardized Force Majeure clauses covering events beyond the reasonable control of either party, including but not limited to: civil unrest, government expropriation, pandemic-related supply chain disruptions, and extreme weather events. In the event of prolonged Force Majeure (exceeding 180 days), both parties will have the right to terminate the affected agreement without penalty, subject to a structured wind-down period.
